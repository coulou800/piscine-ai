### Exercise 3: Embeddings 2

##### 1. Compute the embedding of all the words in this sentence

```python
#install the required spacy model (pipeline)
!python -m spacy download en_core_web_md
```

```python
# from scipy.spatial.distance import cosine
from sklearn.metrics.pairwise import cosine_distances
import seaborn as sns
import matplotlib.pyplot as plt
import sklearn

nlp = sp.load("en_core_web_md")

# Define the sentence
sentence = "laptop computer coffee tea water liquid dog cat kitty"
doc = nlp(sentence)
words = [token.text for token in doc]
reversed_words = words[::-1]

# Compute embeddings
embeddings = [token.vector for token in doc]

cosine_dist = cosine_distances(embeddings)
cosine_dist_reversed = cosine_dist[::-1]
print(sklearn.__version__)
# Create a heatmap
plt.figure(figsize=(10, 4))
sns.heatmap(cosine_dist_reversed, xticklabels=words, yticklabels=reversed_words, cmap='plasma',vmin=0,vmax=1)
plt.title('Cosine distance between words')
plt.show()
```

### Exercise 4: Sentences' similarity

### Exercise 5: NER

```python

text = """Apple Inc. is an American multinational technology company headquartered in Cupertino, California, that designs, develops, and sells consumer electronics, computer software, and online services. It is considered one of the Big Five companies in the U.S. information technology industry, along with Amazon, Google, Microsoft, and Facebook.
Apple was founded by Steve Jobs, Steve Wozniak, and Ronald Wayne in April 1976 to develop and sell Wozniak's Apple I personal computer, though Wayne sold his share back within 12 days. It was incorporated as Apple Computer, Inc., in January 1977, and sales of its computers, including the Apple I and Apple II, grew quickly."""

nlp = sp.load("en_core_web_md")
doc = nlp(text)

entities = [(ent.text, ent.label_) for ent in doc.ents]
print(entities)


```

##### 2.  Run the NER on this sentence and print the named entity, the start_char, the end_char and the label of the named entity.

```python
text = " Paul eats an apple while watching a movie on his Apple device. "

nlp = sp.load("en_core_web_md")
doc = nlp(text)

for ent in doc.ents:
    print(f"Named Entity: {ent.text}, Start Char: {ent.start_char}, End Char: {ent.end_char}, Label: {ent.label_}")
```

### Exercise 6: Part-of-speech tags

```python
with open('data/news_amazon.txt', 'r') as f:
    text = f.read()

nlp = sp.load("en_core_web_md")
doc = nlp(text)

for sent in doc.sents:
    if any(token.text == "Bezos" and token.tag_ == "NNP" for token in sent):
        print(f"INFO:  Bezos PROPN NNP\nSentence:  {sent.text}")
```
